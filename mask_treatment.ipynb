```python
# Steg 1: Importer Image-modulen fra Pillow-biblioteket
from PIL import Image

def process_mask_image(input_filename, output_filename):
    """
    Konverterer den hvite masken i et bilde til en svart silhuett
    med gjennomsiktig bakgrunn.

    Args:
        input_filename (str): Filnavnet til bildet som skal behandles.
        output_filename (str): Filnavnet for det ferdige bildet (bør være .png).
    """
    try:
        # Steg 2: Last inn originalbildet
        # .convert("RGB") sikrer at vi jobber med standard fargekanaler
        # selv om originalen er i et annet format.
        original_image = Image.open(input_filename).convert("RGB")
        print(f"Bildet '{input_filename}' ble lastet inn.")

        # Steg 3: Lag et nytt bilde for resultatet
        # Modus "RGBA" (Rød, Grønn, Blå, Alpha) er nødvendig for gjennomsiktighet.
        # (0, 0, 0, 0) betyr at alle piksler starter som helt gjennomsiktige.
        output_image = Image.new("RGBA", original_image.size, (0, 0, 0, 0))
        print("Opprettet et nytt, gjennomsiktig bilde for resultatet.")

        # Steg 4: Hent pikseldata for effektiv behandling
        input_pixels = original_image.load()
        output_pixels = output_image.load()
        
        width, height = original_image.size
        
        # Definerer en terskel for hva vi anser som "hvit".
        # Dette gjør koden mer robust mot små variasjoner i fargen (f.eks. 254, 255, 255).
        white_threshold = 250 

        # Steg 5: Gå gjennom hver piksel i originalbildet
        print("Behandler piksler...")
        for x in range(width):
            for y in range(height):
                # Hent fargeverdiene (R, G, B) for den aktuelle pikselen
                r, g, b = input_pixels[x, y]
                
                # Steg 6: Sjekk om pikselen er hvit
                if r > white_threshold and g > white_threshold and b > white_threshold:
                    # Hvis den er hvit, sett den tilsvarende pikselen i resultatbildet
                    # til å være svart (0, 0, 0) og helt ugjennomsiktig (255 i alpha-kanalen).
                    output_pixels[x, y] = (0, 0, 0, 255)
        
        # Steg 7: Lagre det nye bildet
        # Vi bruker PNG-formatet fordi det støtter gjennomsiktighet.
        output_image.save(output_filename, "PNG")
        print(f"Ferdig! Resultatet er lagret som '{output_filename}'.")

    except FileNotFoundError:
        print(f"FEIL: Fant ikke filen '{input_filename}'. Sjekk at filnavnet er riktig.")
    except Exception as e:
        print(f"En uventet feil oppstod: {e}")

# --- Slik bruker du koden ---
if __name__ == "__main__":
    # Bytt ut disse filnavnene med dine egne
    input_file = r".\02_DinoV3\data\foreground_segmentation_labels\label_29.png"
    output_file = r".\02_DinoV3\data\foreground_segmentation_labels\label_19.png"
    
    process_mask_image(input_file, input_file)
```
